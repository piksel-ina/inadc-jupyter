# Introduction to the Open Data Cube

## A. Background

The **Open Data Cube** is open-source software for managing and analysing large amounts of Earth observation data. It is released under the Apache 2.0 license and provides a Python interface for querying and loading that data. The Piksel Sandbox is built on the Open Data Cube.

Satellites observe the same locations repeatedly, producing large collections of imagery over time. The Open Data Cube records these in a database and indexes them, which makes searching and retrieval efficient. Rather than locating and opening individual files, the required data is requested and the datacube returns it.

The data is organised along three dimensions: **time**, **longitude**, and **latitude**. This arrangement allows data to be selected by location and date, for example a single area over a single year.

The datacube is accessed from Python through a single object, `dc`. It is used to list the data that is available and to load the portion required.

## B. Description

It covers:
* connecting to the datacube from Python
* listing the products that are available
* inspecting the measurements that a product contains

## C. Connecting to the datacube

A connection is created by constructing a `Datacube` object. The `app` argument is a short label that identifies the session in the database logs.

`dc` is the connection to the datacube. It is used to list the available data and to load it.

```python
import datacube

dc = datacube.Datacube(app="02_odc_introduction")
```

## D. Listing the available products

The `list_products` method returns a table of the products held in the datacube. Each row is a product, which is a collection of satellite data of a single type.

The product used throughout the beginner notebooks is `s2_geomad_annual`. It is an annual summary of Sentinel-2 imagery: a single cloud-free image for each year, derived from all observations collected during that year.

```python
products = dc.list_products()
products
```

## E. Inspecting a product's measurements

A product is made up of measurements. Measurements are the individual bands a product provides, such as red, green, blue, and near-infrared. These are the variables selected when loading data.

The `list_measurements` method returns the measurements for every product. The result below is limited to `s2_geomad_annual`.

```python
measurements = dc.list_measurements()
measurements.loc["s2_geomad_annual"]
```
