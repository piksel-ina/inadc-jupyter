## B. The Open Data Cube

The terms below describe distinct parts of the system, although some of their names are similar.

Open Data Cube (ODC)
: An open-source Python library for managing satellite-imagery archives. It simplifies access, processing, and analysis of decades of geographical data.

Data cube
: A multi-dimensional data structure. In Earth Observation, it organises satellite imagery by dimensions such as time × longitude × latitude × measurement.

`Datacube` class
: The main Python interface to ODC. A `Datacube` object connects code to the ODC index and provides methods for finding and loading data.

Product
: A collection of datasets that share the same set of measurements and some subset of metadata. Landsat Surface Reflectance, Landsat Surface Temperature, and Sentinel-2 imagery are examples of products.

Measurement
: A named variable within a product, with its own data type and units. Red, green, blue, and near-infrared are examples of measurements in a Sentinel-2 imagery product.

## C. Connecting to the datacube

Every ODC operation, including finding or loading data, starts with a `Datacube` object.
Calling `Datacube(app=...)` creates this object and connects it to the ODC index.
The `app` argument gives the application a name; here, that name is `odc_introduction`.
The variable `dc` stores the object for the operations that follow.

```python
from datacube import Datacube

dc = Datacube(app="odc_introduction")
```

## D. Listing the available products

The `list_products` method returns a table of the products in the datacube.

```python
products = dc.list_products()
products
```

> The product used throughout the beginner notebooks is `s2_geomad_annual`: an annual geometric-median composite (a *geomad*) of Sentinel-2 imagery. A geomad combines all observations from the year into a single representative image, down-weighting cloud-contaminated pixels so the result is mostly cloud-free.

## E. Inspecting a product's measurements

The `list_measurements` method returns a table with every measurement of every product in the datacube. Pandas' `.loc` filters that table to just the rows for `s2_geomad_annual`.

```python
measurements = dc.list_measurements()
measurements.loc["s2_geomad_annual"]
```
