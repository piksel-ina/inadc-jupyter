## B. The Jupyter environment

JupyterLab
: The web interface for viewing, editing, and running notebooks.
  Its toolbar, file browser, tabs, and status indicators provide the workspace around a notebook.

Notebook
: A file (`.ipynb`) made up of cells.
  The file open here is itself a notebook.

Cell
: A block of Python code or formatted text within a notebook.
  Each cell runs separately.

Kernel
: The Python process that runs code cells.
  The notebook file stores cells and their most recent output, while the kernel keeps variables and other state between cell runs.
  Each open notebook has its own kernel.

## C. Cell types

A notebook has three kinds of cells: **Code**, **Markdown**, and **Raw**.
Choose a type from the cell-type menu in the toolbar above the notebook.

### 1. Code cells

Code cells contain Python statements that the kernel can run.

The cell below is a Code cell:

`````python
print("This is a code cell")
`````

### 2. Markdown cells

Markdown cells contain formatted text that explains the code around them.
Double-click a Markdown cell to edit its source, then run it to display the formatted result.

> A syntax reference is available from **"Help" → "Markdown Reference"**.

For example, a Markdown cell with this source:

````
**Bold**, _italic_, `inline code`, and a list:

- one
- two
````

renders as:

**Bold**, _italic_, `inline code`, and a list:

- one
- two

### 3. Raw cells

Raw cells contain text that Jupyter does not process.
When a notebook is converted to another format, such as HTML or LaTeX, this text passes through unchanged.
Raw cells are not used elsewhere in this guide.

JupyterLab also shows whether it is busy and provides controls to interrupt, restart, or shut it down.

### 2. Kernel state

The kernel keeps variables in memory between cell runs.
A variable created in one cell therefore remains available to later cells.

`````python
area_name = "Study Area"
`````

`````python
print(f"The value of area_name is: {area_name}")
`````

Cells can run in any order and can run more than once.
As a result, the state held by the kernel may no longer match the order shown on the page, and a notebook that appears complete can still fail when run from top to bottom with a fresh kernel.

:::{tip}
Use **"Kernel" → "Restart Kernel and Run All Cells"** to check that a notebook runs correctly from a clean state.
:::

### 4. Managing the kernel

Use the following actions to stop running code, clear the kernel's state, or release its resources.

| Action        | What it does                                                                 | When to use                                                              | Where                                                                             |
| ------------- | ---------------------------------------------------------------------------- | ------------------------------------------------------------------------ | --------------------------------------------------------------------------------- |
| **Interrupt** | Stops the running cell but keeps the kernel's state.                          | A cell is taking too long or has entered an unintended loop.             | &#9632; **"Stop"** button in the toolbar, or **"Kernel" → "Interrupt Kernel"**. |
| **Restart**   | Clears all kernel state and starts a fresh Python process.                    | The kernel is unresponsive, or a clean top-to-bottom run needs checking. | &#8635; **"Restart"** button in the toolbar, or **"Kernel" → "Restart Kernel"**. |
| **Shutdown**  | Stops the kernel; the notebook remains open without a kernel attached.        | A notebook is no longer active and its memory and compute can be freed.  | **"Running Terminals and Kernels"** panel → select &#10005; beside the notebook. |

The cell below pauses for 20 seconds.
While it is running, select the &#9632; **"Stop"** button or **"Kernel" → "Interrupt Kernel"** to interrupt it:

`````python
import time

time.sleep(20)
`````

## F. Working with notebook files

### 1. Opening, creating, and renaming

Open notebooks from the file browser on the left of JupyterLab.
Double-click a notebook file (`.ipynb`) to open it in a new tab.
If the file browser is hidden, select the &#128193; **"File browser"** icon in the top-left corner to show it again.

To create a notebook, first open its destination directory in the file browser, then select the &#10010; **"New Launcher"** button above the browser.
The Launcher lists the available notebook types and utilities.
Under **"Notebook"**, select **"Python 3"** to create `Untitled.ipynb` in the open directory.

To rename a notebook, right-click it in the file browser and select **"Rename"**.

### 2. Saving

JupyterLab saves changes automatically every few minutes.
To save manually, select **"File" → "Save Notebook"** or use the &#128190; **"Save"** button on the left of the toolbar.

### 3. Exporting

Exporting creates a file that can be opened outside JupyterLab.
Select **"File" → "Save and Export Notebook As" → "Executable Script"** to export the notebook as a `.py` file.
The same submenu also provides HTML, PDF, and Markdown formats.

## G. Useful keyboard shortcuts

JupyterLab keyboard shortcuts depend on the cell mode.
In **command mode**, the cell is selected but not being edited; in **edit mode**, the cursor is inside the cell.
Most shortcuts below require command mode, so press {kbd}`Esc` to leave edit mode first.

| Shortcut                          | Action                                                            |
| --------------------------------- | ----------------------------------------------------------------- |
| {kbd}`Esc`                        | Switch to command mode.                                           |
| {kbd}`Enter`                      | Switch to edit mode.                                              |
| {kbd}`Shift+Enter`                | Run the current cell and move to the next; works in either mode.  |
| {kbd}`Y` / {kbd}`M` / {kbd}`R`    | Set the selected cell to Code, Markdown, or Raw.                  |
| {kbd}`A` / {kbd}`B`               | Insert a new cell above or below the selected cell.               |
| {kbd}`D` {kbd}`D`                 | Delete the selected cell by pressing {kbd}`D` twice in succession. |
| {kbd}`Z`                          | Undo the last cell operation.                                     |
| {kbd}`H`                          | Open the full keyboard shortcut reference.                        |
