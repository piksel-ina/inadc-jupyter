## B. The Jupyter environment

JupyterLab
: The web-based interface for viewing, editing, and running notebooks. The toolbar, file browser, tabs, and status indicators all belong to JupyterLab.

Notebook
: A file (`.ipynb`) made up of cells. The current file is itself a notebook.

Cell
: A block of either Python code or formatted text within a notebook. Cells are run one at a time.

Kernel
: The Python process that runs the code in code cells. The kernel is separate from the notebook file: the notebook stores cells and their most recent output, while the kernel holds variables and state across cell executions. Each open notebook has its own kernel.

## C. Cell types

A notebook has three kinds of cells: **Code**, **Markdown**, and **Raw**. The type is set from the cell-type menu in the toolbar above the notebook.

### 1. Code cells

Code cells hold Python statements.

The cell below is a code cell:


`````python
print("This is a code cell")
`````

### 2. Markdown cells

Markdown cells hold formatted text that documents the code around them. Double-clicking a Markdown cell opens it for editing; running it renders the Markdown.

> A reference for the available syntax is in the **Help** menu under **Markdown Reference**

A Markdown cell with the following source:

````
**Bold**, _italic_, `inline code`, and a list:

- one
- two
````

renders as:


**Bold**, _italic_, `inline code`, and a list:

- one
- two

### 3. Raw cells

Raw cells hold unprocessed text. Their contents are passed through unchanged when the notebook is converted to another format such as HTML or LaTeX. Raw cells are not used elsewhere in this guide.

### 2. Kernel state

The kernel keeps variables in memory across cell executions. A variable set in one cell stays available to later cells.

`````python
area_name = "Lake Toba"
`````

`````python
print(f"The value of area_name is: {area_name}")
`````

Because cells can be run in any order and more than once, the state held by the kernel may not match the order of cells on the page. A notebook that looks complete can still produce errors when run top-to-bottom on a fresh kernel.

:::{tip}
To confirm a notebook runs correctly from a clean state, use **Kernel → Restart Kernel and Run All Cells**.
:::

### 4. Managing the kernel

Three actions manage the kernel directly: interrupting a running cell, restarting the kernel, and shutting it down.

| Action        | What it does                                                                 | When to use                                                              | Where                                                                             |
| ------------- | ---------------------------------------------------------------------------- | ------------------------------------------------------------------------ | --------------------------------------------------------------------------------- |
| **Interrupt** | Stops the running cell; kernel state is kept.                                | A cell is taking too long or has entered an unintended loop.             | &#9632; **Stop** button in toolbar, or **Kernel → Interrupt Kernel**.             |
| **Restart**   | Wipes all kernel state and starts a fresh Python process.                    | The kernel is unresponsive, or to confirm a clean top-to-bottom run.     | &#8635; **Restart** button in toolbar, or **Kernel → Restart Kernel**.            |
| **Shutdown**  | Terminates the kernel; the notebook file stays open with no kernel attached. | Free memory and compute held by a notebook that is no longer active.     | **Running Terminals and Kernels** panel → click &#10005; beside the notebook.     |

The cell below pauses for 20 seconds. Clicking the &#9632; **Stop** button (or selecting **Kernel → Interrupt Kernel**) interrupts it:


`````python
import time

time.sleep(20)
`````

## F. Working with notebook files

### 1. Opening, creating, and renaming

Notebooks are opened from the file browser on the left of JupyterLab. Double-clicking a notebook file (`.ipynb`) opens it in a new tab. If the file browser is not visible, the &#128193; **File browser** icon at the top-left of the screen reopens it.

To create a new notebook, navigate to the target directory in the file browser and click the &#10010; **New Launcher** button above it. The Launcher lists the available notebook types and utilities; the **Python 3** button under the **Notebook** heading creates a new notebook named `Untitled.ipynb` in the current directory.

Right-clicking a notebook in the file browser and selecting **Rename** changes its name.

### 2. Saving

Modifications are saved automatically every few minutes. A manual save is available from **File → Save Notebook**, or via the &#128190; **Save** button on the left of the toolbar.

### 3. Exporting

A notebook can be exported to formats that open outside JupyterLab. Use **File → Save and Export Notebook As → Executable Script** to export as a `.py` file. Other export formats (HTML, PDF, Markdown) are available in the same submenu.

## G. Useful keyboard shortcuts

Keyboard shortcuts in JupyterLab depend on whether the cell is in **command mode** (selected, not editing) or **edit mode** (cursor inside). Most of the shortcuts below require command mode; press {kbd}`Esc` to switch out of edit mode first.

| Shortcut                          | Action                                                            |
| --------------------------------- | ----------------------------------------------------------------- |
| {kbd}`Esc`                        | Switch to command mode.                                           |
| {kbd}`Enter`                      | Switch to edit mode.                                              |
| {kbd}`Shift+Enter`                | Run the current cell and move to the next. Works in either mode.  |
| {kbd}`Y` / {kbd}`M` / {kbd}`R`    | Set the selected cell to Code, Markdown, or Raw type.             |
| {kbd}`A` / {kbd}`B`               | Insert a new cell above or below the selected one.                |
| {kbd}`D` {kbd}`D`                 | Delete the selected cell (press {kbd}`D` twice in succession).    |
| {kbd}`Z`                          | Undo the last cell operation.                                     |
| {kbd}`H`                          | Open the full keyboard shortcut reference.                        |
