## B. Open Data Cube

Istilah berikut merujuk pada bagian sistem yang berbeda meskipun beberapa namanya mirip.

Open Data Cube (ODC)
: Pustaka Python sumber terbuka untuk mengelola arsip citra satelit.
  Pustaka ini menyederhanakan akses, pemrosesan, dan analisis arsip data geografis yang mencakup rentang puluhan tahun.

Kubus data
: Struktur data multidimensi.
  Dalam pengamatan Bumi, struktur ini menata citra satelit berdasarkan dimensi seperti waktu × garis bujur × garis lintang × measurement.

Class `Datacube`
: Antarmuka utama untuk mengakses ODC melalui Python.
  Objek `Datacube` menghubungkan kode ke indeks ODC dan menyediakan metode untuk mencari serta memuat data.

Produk
: Kumpulan dataset dengan rangkaian measurement yang sama dan sebagian metadata yang sama.
  Landsat Surface Reflectance, Landsat Surface Temperature, dan citra Sentinel-2 merupakan contoh produk.

Measurement
: Variabel bernama dalam suatu produk, dengan tipe data dan unitnya sendiri.
  Merah, hijau, biru, dan inframerah dekat merupakan contoh measurement dalam produk citra Sentinel-2.

## C. Koneksi ke datacube

Setiap operasi pada ODC, termasuk pencarian dan pemuatan data, dimulai dengan membuat objek dari class `Datacube`.
Pemanggilan `Datacube(app=...)` membuat objek tersebut dan menghubungkannya ke indeks ODC.
Argumen `app` menetapkan nama aplikasi; dalam notebook ini, namanya adalah `odc_introduction`.
Variabel `dc` menyimpan objek tersebut untuk operasi berikutnya.

```python
from datacube import Datacube

dc = Datacube(app="odc_introduction")
```

## D. Daftar produk yang tersedia

Metode `list_products` menghasilkan tabel produk dalam datacube.

```python
products = dc.list_products()
products
```

> Produk yang digunakan di seluruh notebook Panduan Pemula adalah `s2_geomad_annual`, yaitu komposit median geometris tahunan (sebuah *geomad*) dari citra Sentinel-2.
> Geomad menggabungkan seluruh observasi selama satu tahun menjadi satu citra representatif dan mengurangi bobot piksel yang terkontaminasi awan sehingga hasilnya sebagian besar bebas awan.

## E. Pemeriksaan measurement dalam suatu produk

Metode `list_measurements` menghasilkan tabel yang memuat setiap measurement dari setiap produk dalam datacube.
`.loc` milik Pandas memfilter tabel tersebut agar hanya memuat baris untuk `s2_geomad_annual`.

```python
measurements = dc.list_measurements()
measurements.loc["s2_geomad_annual"]
```
