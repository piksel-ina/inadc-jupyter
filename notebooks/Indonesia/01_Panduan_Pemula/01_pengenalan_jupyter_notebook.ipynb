## B. Lingkungan Jupyter

JupyterLab
: Antarmuka web untuk membuka, menyunting, dan menjalankan notebook.
  Bilah alat, peramban file, tab, dan indikator status membentuk ruang kerja di sekitar notebook.

Notebook
: File (`.ipynb`) yang tersusun atas sejumlah sel.
  Materi yang sedang dibuka ini juga berupa notebook.

Sel
: Blok berisi kode Python atau teks berformat di dalam notebook.
  Setiap sel dijalankan secara terpisah.

Kernel
: Proses Python yang menjalankan sel Code.
  File notebook menyimpan sel beserta hasil terakhirnya, sedangkan kernel menyimpan variabel dan keadaan lain di antara eksekusi sel.
  Setiap notebook yang terbuka memiliki kernel tersendiri.

(cell-types)=
## C. Jenis sel

Notebook memiliki tiga jenis sel: **Code**, **Markdown**, dan **Raw**.
Pilih jenis sel melalui menu pada bilah alat di atas notebook.

### 1. Sel Code

Sel Code berisi pernyataan Python yang dapat dijalankan oleh kernel.

Sel berikut merupakan sel Code:

`````python
print("This is a code cell")
`````

### 2. Sel Markdown

Sel Markdown berisi teks berformat untuk menjelaskan kode di sekitarnya.
Klik dua kali sel Markdown untuk menyunting sumbernya, lalu jalankan sel tersebut untuk menampilkan hasil yang sudah diformat.

> Referensi sintaks tersedia melalui **"Help" → "Markdown Reference"**.

Sebagai contoh, tulis sumber berikut dalam sel Markdown:

````
**Tebal**, _miring_, `kode sebaris`, dan sebuah daftar:

- satu
- dua
````

Hasilnya akan tampil seperti ini:

**Tebal**, _miring_, `kode sebaris`, dan sebuah daftar:

- satu
- dua

### 3. Sel Raw

Sel Raw memuat teks yang tidak diproses oleh Jupyter.
Saat notebook dikonversi ke format lain, seperti HTML atau LaTeX, teks tersebut diteruskan tanpa perubahan.
Sel Raw tidak digunakan lagi dalam panduan ini.

## D. Kernel

Kernel menjalankan kode dan menyimpan keadaannya di antara eksekusi sel.
JupyterLab juga menampilkan apakah kernel sedang sibuk serta menyediakan kontrol untuk menginterupsi, memulai ulang, atau mematikannya.

### 1. Menjalankan sel

Tekan {kbd}`Shift+Enter` atau pilih tombol &#9654; **"Run"** pada bilah alat notebook untuk menjalankan sel yang dipilih.
Hasilnya akan muncul tepat di bawah sel.

Untuk menjalankan semua sel dari atas ke bawah, pilih **"Run" → "Run All Cells"** pada bilah menu atas.

Jalankan sel berikut untuk mencetak pesan singkat:

`````python
print("Cell executed.")
`````

### 2. Keadaan kernel

Kernel menyimpan variabel dalam memori di antara eksekusi sel.
Karena itu, variabel yang dibuat dalam satu sel tetap dapat digunakan oleh sel berikutnya.

`````python
area_name = "Study Area"
`````

`````python
print(f"The value of area_name is: {area_name}")
`````

Sel dapat dijalankan dalam urutan apa pun dan lebih dari sekali.
Akibatnya, keadaan di dalam kernel mungkin tidak lagi sesuai dengan urutan sel pada halaman, dan notebook yang tampak lengkap tetap dapat menghasilkan galat saat dijalankan dari atas ke bawah dengan kernel baru.

:::{tip}
Pilih **"Kernel" → "Restart Kernel and Run All Cells"** untuk memastikan notebook dapat dijalankan dengan benar dari keadaan bersih.
:::

### 4. Mengelola kernel

Gunakan tindakan berikut untuk menghentikan kode, menghapus keadaan kernel, atau membebaskan sumber daya yang dipakainya.

| Tindakan      | Fungsi                                                                        | Kapan digunakan                                                           | Lokasi                                                                            |
| ------------- | ----------------------------------------------------------------------------- | ------------------------------------------------------------------------- | --------------------------------------------------------------------------------- |
| **Interrupt** | Menghentikan sel yang sedang berjalan tanpa menghapus keadaan kernel.         | Saat sel berjalan terlalu lama atau memasuki perulangan yang tidak disengaja. | Tombol &#9632; **"Stop"** pada bilah alat, atau **"Kernel" → "Interrupt Kernel"**. |
| **Restart**   | Menghapus seluruh keadaan kernel dan memulai proses Python yang baru.          | Saat kernel tidak merespons, atau untuk memeriksa eksekusi bersih dari atas ke bawah. | Tombol &#8635; **"Restart"** pada bilah alat, atau **"Kernel" → "Restart Kernel"**. |
| **Shutdown**  | Menghentikan kernel; notebook tetap terbuka tanpa kernel yang terhubung.       | Saat notebook tidak lagi aktif sehingga memori dan komputasinya dapat dibebaskan. | Panel **"Running Terminals and Kernels"** → pilih &#10005; di samping notebook. |

Sel berikut berhenti selama 20 detik.
Saat sel masih berjalan, pilih tombol &#9632; **"Stop"** atau **"Kernel" → "Interrupt Kernel"** untuk menginterupsinya:

`````python
import time

time.sleep(20)
`````

## F. Mengelola file notebook

### 1. Membuka, membuat, dan mengganti nama

Buka notebook melalui peramban file di sisi kiri JupyterLab.
Klik dua kali file notebook (`.ipynb`) untuk membukanya di tab baru.
Jika peramban file tidak terlihat, pilih ikon &#128193; **"File browser"** di sudut kiri atas untuk menampilkannya kembali.

Untuk membuat notebook, buka dahulu direktori tujuan di peramban file, lalu pilih tombol &#10010; **"New Launcher"** di atas peramban.
Launcher menampilkan jenis notebook dan utilitas yang tersedia.
Di bawah **"Notebook"**, pilih **"Python 3"** untuk membuat `Untitled.ipynb` dalam direktori yang sedang dibuka.

Untuk mengganti nama notebook, klik kanan file tersebut dalam peramban file lalu pilih **"Rename"**.

### 2. Menyimpan

JupyterLab menyimpan perubahan secara otomatis setiap beberapa menit.
Untuk menyimpan secara manual, pilih **"File" → "Save Notebook"** atau gunakan tombol &#128190; **"Save"** di sisi kiri bilah alat.

### 3. Mengekspor

Ekspor menghasilkan file yang dapat dibuka di luar JupyterLab.
Pilih **"File" → "Save and Export Notebook As" → "Executable Script"** untuk mengekspor notebook sebagai file `.py`.
Submenu yang sama juga menyediakan format HTML, PDF, dan Markdown.

## G. Pintasan papan ketik

Pintasan papan ketik JupyterLab bergantung pada mode sel.
Dalam **command mode**, sel sedang dipilih tetapi tidak disunting; dalam **edit mode**, kursor berada di dalam sel.
Sebagian besar pintasan berikut memerlukan command mode, jadi tekan {kbd}`Esc` untuk keluar dari edit mode terlebih dahulu.

| Pintasan                          | Tindakan                                                          |
| --------------------------------- | ----------------------------------------------------------------- |
| {kbd}`Esc`                        | Beralih ke command mode.                                          |
| {kbd}`Enter`                      | Beralih ke edit mode.                                             |
| {kbd}`Shift+Enter`                | Menjalankan sel aktif dan berpindah ke sel berikutnya; dapat digunakan dalam kedua mode. |
| {kbd}`Y` / {kbd}`M` / {kbd}`R`   | Menetapkan sel yang dipilih sebagai Code, Markdown, atau Raw.     |
| {kbd}`A` / {kbd}`B`              | Menyisipkan sel baru di atas atau di bawah sel yang dipilih.      |
| {kbd}`D` {kbd}`D`                | Menghapus sel yang dipilih dengan menekan {kbd}`D` dua kali secara berurutan. |
| {kbd}`Z`                          | Membatalkan operasi terakhir pada sel.                            |
| {kbd}`H`                          | Membuka referensi lengkap pintasan papan ketik.                   |
